## Conceptual Question 3

The fitted least-squares model predicts starting salary, measured in thousands of dollars, using GPA, IQ, education level, and interaction terms between GPA and IQ and between GPA and education level.

The fitted model is:

$$
\hat{Y}
=
50
+20(GPA)
+0.07(IQ)
+35(Level)
+0.01(GPA \times IQ)
-10(GPA \times Level)
$$

where $Level = 1$ for college graduates and $Level = 0$ for high school graduates.

### Part (a)

The effect of education level depends on GPA because the model contains a GPA-by-Level interaction. Holding GPA and IQ constant, the predicted difference between college and high school graduates is:

$$
35 - 10(GPA)
$$

The difference equals zero at a GPA of 3.5. For GPA values greater than 3.5, the difference becomes negative, meaning that the model predicts a higher starting salary for high school graduates. Therefore, **option (iii)** is correct: high school graduates are predicted to earn more than college graduates when GPA is sufficiently high.

### Part (b)

For a college graduate with a GPA of 4.0 and an IQ of 110:

$$
\hat{Y}
=
50+20(4.0)+0.07(110)+35+0.01(4.0)(110)-10(4.0)
$$

$$
\hat{Y}=137.1
$$

Because salary is measured in thousands of dollars, the predicted starting salary is **$137,100**.

### Part (c)

The statement is **false**. The numerical size of an interaction coefficient does not by itself indicate whether the interaction effect is weak or statistically unimportant. The effect of the interaction depends on both the coefficient and the scale of the variables being multiplied.

For example, for an individual with a GPA of 4.0 and an IQ of 110, the GPA-by-IQ interaction contributes:

$$
0.01(4.0)(110)=4.4
$$

Because salary is measured in thousands of dollars, this represents a contribution of $4,400 to the predicted salary. Therefore, a coefficient that appears small can still have a meaningful effect when the interacting variables take relatively large values.

Statistical evidence for an interaction should instead be evaluated using measures such as the coefficient's standard error, confidence interval, or p-value. The coefficient magnitude alone is insufficient to determine whether there is little evidence of an interaction.

## Applied Question 10: Carseats Regression Analysis

The Carseats dataset is used to examine the relationship between product sales and three predictors: Price, Urban, and US. Sales is the quantitative response variable, Price is quantitative, and Urban and US are qualitative Yes/No variables.

The first model estimates Sales as a function of all three predictors using ordinary least squares regression.

```python
from sklearn import datasets
from ISLP import load_data
import statsmodels.api as sm
from statsmodels.formula.api import ols
import pandas as pd

# Load the Carseat dataset
Carseats = load_data("Carseats")
Carseats.head()
```

```python
# Review the variables used in the regression model
Carseats[["Sales", "Price", "Urban", "US"]].info()
```

```python
#Confirm the categories represented by the qualitative predictors
print(Carseats["Urban"].value_counts())
print(Carseats["US"].value_counts())
```

```python
# Check for missing values in the analysis variables
Carseats[["Sales", "Price", "Urban", "US"]].isna().sum()
```

```python
# Fit the multiple regression model to predict Sales using Price, Urban, and US
model_1 = ols("Sales ~ Price + Urban + US", data = Carseats).fit()

model_1.summary()
```

### Part (b): Interpretation of Model Coefficients

The fitted regression model estimates Sales as a function of Price, Urban location, and U.S. location. Statistical significance was evaluated using an alpha level of 0.05.

- **Intercept = 13.0435,** $p < .001$:  
  The intercept represents the predicted Sales value when Price equals 0 and both qualitative variables are at their reference level (`Urban = No` and `US = No`). The intercept is statistically significant, although a price of 0 is not practically meaningful in this context.

- **Price = -0.0545,** $p < .001$:  
  Holding Urban and U.S. location constant, each $1 increase in Price is associated with a decrease of approximately 0.0545 thousand units in Sales, or about 54.5 fewer units sold on average. Price is statistically significant, providing strong evidence of an association between Price and Sales.

- **Urban[T.Yes] = -0.0219,** $p = .936$:  
  Holding Price and U.S. location constant, urban stores are predicted to sell approximately 0.0219 thousand fewer units than non-urban stores. This difference is not statistically significant, indicating no evidence of an association between Urban status and Sales after controlling for the other predictors.

- **US[T.Yes] = 1.2006,** $p < .001$:  
  Holding Price and Urban status constant, stores located in the U.S. are predicted to sell approximately 1.2006 thousand more units than stores outside the U.S., or about 1,201 additional units on average. The coefficient is statistically significant, providing strong evidence of an association between U.S. location and Sales.

### Part (c): Regression Equation

The fitted regression model is:

$$
\widehat{Sales}
=
13.0435
-0.0545(Price)
-0.0219(Urban_{Yes})
+1.2006(US_{Yes})
$$

where:

- $Urban_{Yes} = 1$ if the store is in an urban location and 0 otherwise.
- $US_{Yes} = 1$ if the store is located in the United States and 0 otherwise.

The reference categories are therefore `Urban = No` and `US = No`.

### Part (d): Predictor Significance

Statistical significance was evaluated using an alpha level of 0.05.

- **Price:** $p < .001$, so the null hypothesis $H_0:\beta_{Price}=0$ is rejected.
- **Urban:** $p = .936$, so the null hypothesis $H_0:\beta_{Urban}=0$ is not rejected.
- **US:** $p < .001$, so the null hypothesis $H_0:\beta_{US}=0$ is rejected.

Therefore, **Price and US show evidence of association with Sales**, while Urban does not. Based on these results, Price and US should be retained in the reduced model and Urban should be removed.

```python
# Fit the multiple regression model to predict Sales using Price and US
model_2 = ols("Sales ~ Price + US", data = Carseats).fit()

model_2.summary()
```

### Part (e): Reduced Regression Model

Because Urban was not statistically significant in Model 1, a reduced model was fit using only Price and US.

The fitted reduced model is:

$$
\widehat{Sales}
=
13.0308
-0.0545(Price)
+1.1996(US_{Yes})
$$

Both retained predictors remain statistically significant at $p < .001$. Removing Urban produced virtually no change in model fit, suggesting that Urban did not contribute meaningful explanatory information after accounting for Price and US.

### Part (f): Comparison of Model Fit

Model 1 and Model 2 produced nearly identical overall fit. Both models had an $R^2$ of 0.239, indicating that approximately 23.9% of the variability in Sales is explained by the included predictors. This suggests that both models provide only a modest fit to the data.

The adjusted $R^2$ increased slightly from 0.234 in Model 1 to 0.235 in Model 2 after Urban was removed. Model 2 also produced lower AIC and BIC values, indicating a preferable balance between model fit and complexity.

Because removing Urban did not reduce explanatory performance, the reduced model is more parsimonious and is preferred over the full model.

```python
model_2.conf_int(alpha=0.05)
```

### Part (g): 95% Confidence Intervals

The 95% confidence intervals for the coefficients in Model 2 were:

- **Intercept:** $[11.79032, 14.271265]$
- **US[T.Yes]:** $[0.69152, 1.707766]$
- **Price:** $[-0.06476, -0.044195]$

The confidence intervals for both Price and US exclude zero, which is consistent with their statistically significant p-values. The Price interval is entirely negative, supporting a negative association between Price and Sales. The US interval is entirely positive, supporting higher average Sales for U.S. stores relative to non-U.S. stores, holding Price constant.

```python
# Examine outliers, leverage, and influence for Model 2
influence = model_2.get_influence()

diagnostics = pd.DataFrame({
    "studentized_residual": influence.resid_studentized_external,
    "leverage": influence.hat_matrix_diag,
    "cooks_distance": influence.cooks_distance[0]
})

diagnostics.head()
```

```python
n = Carseats.shape[0]
p = int(model_2.df_model + 1)   # predictors plus intercept

residual_cutoff = 2
leverage_cutoff = 2 * p / n
cooks_cutoff = 4 / n

print("Leverage cutoff:", leverage_cutoff)
print("Cook's distance cutoff:", cooks_cutoff)
```

```python
print("Potential outliers:", (diagnostics["studentized_residual"].abs() > residual_cutoff).sum())

print("High leverage observations:", (diagnostics["leverage"] > leverage_cutoff).sum())

print("Potentially influential observations:", (diagnostics["cooks_distance"] > cooks_cutoff).sum())
```

### Part (h): Outliers and High-Leverage Observations

Diagnostic measures were examined for Model 2 using studentized residuals, leverage values, and Cook's distance.

Some observations were flagged by the more conservative screening rule of $|\text{studentized residual}| > 2$; however, the studentized residuals remained within approximately ±3. Therefore, there is no evidence of severe outliers in the response variable.

The leverage diagnostics identified observations with leverage values above the expected range, indicating the presence of some high-leverage observations. These cases have unusual combinations of the predictor values relative to the rest of the data and therefore warrant additional examination. Cook's distance was also reviewed to assess whether individual observations exert substantial influence on the fitted model.

Overall, the reduced model does not show evidence of extreme response outliers, but it does contain some high-leverage observations.
